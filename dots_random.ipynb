```python
import random

def generate_random_string(phrase):
    # Define the length of the random characters before and after the phrase
    length = 6
    
    # Create a string of possible characters (either ':' or '.')
    prefix_characters = [random.choice([':', '.']) for _ in range(length)]
    suffix_characters = [random.choice([':', '.']) for _ in range(length)]
    
    # Join the characters to form the prefix and suffix
    prefix = ''.join(prefix_characters)
    suffix = ''.join(suffix_characters)
    
    # Combine everything into the final string
    random_string = f"{prefix} {phrase} {suffix}"
    return random_string

# Example usage
phrase = "XAV COFFEE WORKS"
random_string = generate_random_string(phrase)
print(random_string)
```
